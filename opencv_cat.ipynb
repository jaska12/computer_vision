### 2-Katakcha: Rasmni yuklab olish va ekranga chiqarish

```python
from google.colab.patches import cv2_imshow

# Rasmni o'qish
image = cv2.imread("/content/cat_image.jpg")

# Ekranga chiqarish
cv2_imshow(image)
```

### 3-Katakcha: Rasm hajmi va rang kanallarini tekshirish (`shape`)

```python
print("SHAPE:", image.shape)
# 463 => height
# 703 => width
# 3 => color channels (BGR) => Blue, Green, Red

print("HEIGHT:", image.shape[0])
print("WIDTH:", image.shape[1])
print("CHANNELS:", image.shape[2])
```

### 4-Katakcha: Muayyan pikseldagi rang qiymatlarini ko'rish (BGR)

```python
(b, g, r) = image[100, 50]
print("Blue:", b, "Green:", g, "Red:", r)
```

### 5-Katakcha: Rasmni oq-qora (grayscale) formatga o'tkazish

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

### 6-Katakcha: Rasm o'lchamini o'zgartirish (`resize`)

```python
resized = cv2.resize(gray, (300, 300))
cv2_imshow(resized)
```

### 7-Katakcha: Rasmdan kerakli qismni qirqib olish (`cropping`)

```python
cropped = gray[100:300, 150:350]
cv2_imshow(cropped)
```

### 8-Katakcha: Rasmni burish (`rotate`)

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_COUNTERCLOCKWISE)
cv2_imshow(rotated)
```

### 9-Katakcha: Rasmni oyna shaklida ag'darish (`flip`)

```python
flipped = cv2.flip(gray, 1)
cv2_imshow(flipped)
```

### 10-Katakcha: Yangi rasmni saqlash va qayta o'qish (`imwrite`)

```python
# Qirqib olingan rasmni yangi fayl sifatida saqlash
cv2.imwrite("cropped_cat.jpg", cropped)

# Saqlangan yangi rasmni ochib tekshirish
cropped_cat = cv2.imread("/content/cropped_cat.jpg")
cv2_imshow(cropped_cat)
```
